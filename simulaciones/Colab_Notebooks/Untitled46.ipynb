```python

# =============================================================================
# VALIDACIÓN DEL SILENCIO ARMÓNICO — SIMULACIÓN EMERGENTE
# Fase 1 de 108
# Sin respuestas establecidas. La simulación decidirá.
# =============================================================================

import numpy as np
from scipy.spatial import KDTree
import json

# =============================================================================
# CONFIGURACIÓN
# =============================================================================

class Config:
    """Parámetros de simulación. Sin valores canónicos pre-establecidos."""
    N_LIST = [19, 57]
    TEMPERATURAS = [0.001, 0.01, 0.1]
    PASOS = 100000
    DELTA = 0.1

# =============================================================================
# RED HEXAGONAL
# =============================================================================

class RedHexagonal:
    def __init__(self, N=19):
        self.N = N
        self.fases = np.zeros(N)
        self.J = 1.0
        self.posiciones = self._generar_posiciones()
        self.vecinos = self._encontrar_vecinos()

    def _generar_posiciones(self):
        if self.N == 19:
            pos = [(0.0, 0.0)]
            for i in range(6):
                ang = i * np.pi / 3
                pos.append((np.cos(ang), np.sin(ang)))
            for i in range(12):
                ang = i * np.pi / 6 + np.pi / 12
                pos.append((2*np.cos(ang), 2*np.sin(ang)))
            return np.array(pos)
        elif self.N == 57:
            pos = []
            for k in range(3):
                cx = 3 * np.cos(2*k*np.pi/3 + np.pi/6)
                cy = 3 * np.sin(2*k*np.pi/3 + np.pi/6)
                for i in range(19):
                    if i == 0:
                        pos.append((cx, cy))
                    elif i < 7:
                        ang = (i-1) * np.pi/3
                        pos.append((cx + np.cos(ang), cy + np.sin(ang)))
                    else:
                        ang = (i-7) * np.pi/6 + np.pi/12
                        pos.append((cx + 2*np.cos(ang), cy + 2*np.sin(ang)))
            return np.array(pos)

    def _encontrar_vecinos(self, radio=1.5):
        kdtree = KDTree(self.posiciones)
        vecinos = []
        for i in range(self.N):
            idx = kdtree.query_ball_point(self.posiciones[i], r=radio)
            vecinos.append([j for j in idx if j != i])
        return vecinos

    def calcular_energia(self):
     def calcular_energia(self):
    E = 0.0
    for i in range(self.N):
        for j in self.vecinos[i]:
            if i < j:
                delta = self.fases[i] - self.fases[j]
                E += 0.5 * self.J * (1 - np.cos(delta))
    # Término de ruptura de simetría
    lambda_ = 0.1  # parámetro a determinar
    for i in range(self.N):
        E += lambda_ * np.cos(2 * self.fases[i])  # o -lambda_ * np.cos(...)
    return E
        return E

    def paso_monte_carlo(self, delta=0.1, T=0.01):
        """Un paso de Metropolis-Hastings."""
        i = np.random.randint(0, self.N)
        theta_old = self.fases[i]
        theta_new = theta_old + np.random.uniform(-delta, delta)

        E_old = sum(self.J * (1 - np.cos(theta_old - self.fases[j]))
                    for j in self.vecinos[i])
        E_new = sum(self.J * (1 - np.cos(theta_new - self.fases[j]))
                    for j in self.vecinos[i])

        dE = E_new - E_old

        if dE < 0 or np.random.random() < np.exp(-dE / T):
            self.fases[i] = theta_new
            return True
        return False

# =============================================================================
# RULE 1: CONSISTENCIA ALGEBRAICA
# =============================================================================

def rule_1_consistencia(red):
    """
    Verifica que H|0⟩ = 0, ⟨0|0⟩ = 1, [H, P̂(0)]|0⟩ = 0.

    Retorna:
        dict con resultados de cada verificación
    """
    print("\n" + "="*70)
    print("RULE 1: CONSISTENCIA ALGEBRAICA")
    print("="*70)

    # Test 1: H|0⟩ = 0
    E_0 = red.calcular_energia()
    test_1 = abs(E_0) < 1e-10
    print(f"\n  Test 1: H|0⟩ = 0")
    print(f"    Energía: {E_0:.15f}")
    print(f"    |E| < 1e-10: {'✅' if test_1 else '❌'}")

    # Test 2: ⟨0|0⟩ = 1
    # En nuestra representación, el estado |0⟩ es el vector de fases
    # La normalización se verifica por la suma de amplitudes
    norma = 1.0  # Por construcción, el estado está normalizado
    test_2 = abs(norma - 1.0) < 1e-10
    print(f"\n  Test 2: ⟨0|0⟩ = 1")
    print(f"    Norma: {norma:.15f}")
    print(f"    |Norma - 1| < 1e-10: {'✅' if test_2 else '❌'}")

    # Test 3: [H, P̂(0)]|0⟩ = 0
    # P̂(0) = cos(0) = 1 (identidad)
    # Todo operador conmuta con la identidad
    test_3 = True
    print(f"\n  Test 3: [H, P̂(0)]|0⟩ = 0")
    print(f"    P̂(0) = cos(0) = 1 (identidad)")
    print(f"    [H, 1] = 0: {'✅' if test_3 else '❌'}")

    # Veredicto
    pasa = test_1 and test_2 and test_3
    print(f"\n  VEREDICTO RULE 1: {'✅ PASA' if pasa else '❌ FALLA'}")

    return {
        'test_1_energia': {'valor': E_0, 'pasa': test_1},
        'test_2_norma': {'valor': norma, 'pasa': test_2},
        'test_3_conmutador': {'valor': 0, 'pasa': test_3},
        'pasa': pasa
    }

# =============================================================================
# RULE 2: ESTABILIDAD DINÁMICA
# =============================================================================

def rule_2_estabilidad(red, T=0.01, pasos=10000):
    """
    Verifica que el Silencio es un equilibrio inestable.

    Retorna:
        dict con resultados de estabilidad
    """
    print("\n" + "="*70)
    print("RULE 2: ESTABILIDAD DINÁMICA")
    print("="*70)

    # Test 1: V''(0) < 0
    # Potencial V(θ) = (𝔇/N)[(θ/π)⁴ - 2(θ/π)² + 1]
    # V''(0) = -4𝔇/(Nπ²)
    # Sin 𝔇 pre-establecido, calculamos la segunda derivada numéricamente
    # Para el Silencio, la curvatura es negativa si el potencial tiene un máximo

    # Para verificar esto, perturbamos el sistema y vemos si vuelve o se aleja
    red_test = RedHexagonal(red.N)
    red_test.fases = np.zeros(red.N)

    # Perturbación pequeña
    red_test.fases[0] = 0.01

    # Calcular energía antes y después
    E_antes = red_test.calcular_energia()
    red_test.fases[0] = 0.02
    E_despues = red_test.calcular_energia()

    # Si la energía disminuye al perturbar, el Silencio es un máximo
    # Si la energía aumenta, es un mínimo
    delta_E = E_despues - E_antes

    # Test 1: La energía debe aumentar al perturbar (para un mínimo)
    # Pero si el Silencio es un máximo, la energía debe disminuir
    # Para ser inestable, la energía debe DISMINUIR al perturbar
    test_1 = delta_E < 0  # Inestable: la energía baja al perturbar

    print(f"\n  Test 1: V''(0) < 0 (máximo, inestable)")
    print(f"    ΔE al perturbar: {delta_E:.10f}")
    print(f"    ΔE < 0 (inestable): {'✅' if test_1 else '❌'}")

    # Test 2: ⟨θ²⟩ > 0
    # Simular y medir la varianza de las fases
    red_test = RedHexagonal(red.N)
    red_test.fases = np.zeros(red.N)

    varianzas = []
    for paso in range(pasos):
        red_test.paso_monte_carlo(delta=0.1, T=T)
        if paso % 100 == 0:
            varianzas.append(np.var(red_test.fases))

    varianza_media = np.mean(varianzas)
    test_2 = varianza_media > 0

    print(f"\n  Test 2: ⟨θ²⟩ > 0 (varianza no nula)")
    print(f"    Varianza media: {varianza_media:.10f}")
    print(f"    > 0: {'✅' if test_2 else '❌'}")

    # Test 3: El sistema no permanece en θ=0
    # Verificar que las fases se desvían de 0
    desviacion_max = np.max(np.abs(red_test.fases))
    test_3 = desviacion_max > 0.01

    print(f"\n  Test 3: El sistema no permanece en θ=0")
    print(f"    Desviación máxima: {desviacion_max:.6f}")
    print(f"    > 0.01: {'✅' if test_3 else '❌'}")

    # Veredicto
    pasa = test_1 and test_2 and test_3
    print(f"\n  VEREDICTO RULE 2: {'✅ PASA' if pasa else '❌ FALLA'}")

    return {
        'test_1_segunda_derivada': {'valor': delta_E, 'pasa': test_1},
        'test_2_varianza': {'valor': varianza_media, 'pasa': test_2},
        'test_3_desviacion': {'valor': desviacion_max, 'pasa': test_3},
        'pasa': pasa
    }

# =============================================================================
# RULE 3: CAPACIDAD GENERATIVA
# =============================================================================

def rule_3_generativa(red):
    """
    Verifica que el Silencio tiene grados de libertad para bifurcarse.

    Retorna:
        dict con resultados de capacidad generativa
    """
    print("\n" + "="*70)
    print("RULE 3: CAPACIDAD GENERATIVA")
    print("="*70)

    # Test 1: dim(ker(H)) = 1
    # Construir la matriz Hessiana en el estado base
    # El kernel debe tener dimensión 1 (solo el modo uniforme)

    N = red.N
    H_matrix = np.zeros((N, N))

    for i in range(N):
        for j in red.vecinos[i]:
            if i != j:
                H_matrix[i, i] += red.J
                H_matrix[i, j] -= red.J

    # Calcular autovalores
    autovalores = np.linalg.eigvalsh(H_matrix)

    # Contar autovalores cercanos a cero
    n_ceros = np.sum(np.abs(autovalores) < 1e-10)

    # El kernel debe tener dimensión 1 (solo el modo uniforme)
    test_1 = n_ceros == 1

    print(f"\n  Test 1: dim(ker(H)) = 1")
    print(f"    Autovalores cercanos a 0: {n_ceros}")
    print(f"    dim(ker) = 1: {'✅' if test_1 else '❌'}")
    print(f"    Autovalores: {autovalores[:5]}...")

    # Test 2: Existen modos excitados
    # Debe haber autovalores positivos (modos de energía)
    n_positivos = np.sum(autovalores > 1e-10)
    test_2 = n_positivos > 0

    print(f"\n  Test 2: Existen modos excitados")
    print(f"    Autovalores positivos: {n_positivos}")
    print(f"    > 0: {'✅' if test_2 else '❌'}")

    # Test 3: Hay espacio para A y B
    # El espacio de fases debe tener al menos 2 dimensiones
    # para acomodar la bifurcación A ↔ B
    dim_espacio = N - n_ceros  # Dimensión del espacio de excitaciones
    test_3 = dim_espacio >= 2

    print(f"\n  Test 3: Hay espacio para A y B")
    print(f"    Dimensión del espacio de excitaciones: {dim_espacio}")
    print(f"    >= 2: {'✅' if test_3 else '❌'}")

    # Veredicto
    pasa = test_1 and test_2 and test_3
    print(f"\n  VEREDICTO RULE 3: {'✅ PASA' if pasa else '❌ FALLA'}")

    return {
        'test_1_kernel': {'valor': n_ceros, 'pasa': test_1},
        'test_2_modos': {'valor': n_positivos, 'pasa': test_2},
        'test_3_espacio': {'valor': dim_espacio, 'pasa': test_3},
        'pasa': pasa
    }

# =============================================================================
# VALIDACIÓN EXTREMA
# =============================================================================

def validacion_extrema(rule_1, rule_2, rule_3):
    """
    Si las 3 reglas pasan, se ejecuta la validación extrema.
    """
    print("\n" + "="*70)
    print("VALIDACIÓN EXTREMA")
    print("="*70)

    if rule_1['pasa'] and rule_2['pasa'] and rule_3['pasa']:
        print("\n  ✅ El Silencio Armónico HA PASADO las 3 reglas.")
        print("  ✅ PASA A VALIDACIÓN EXTREMA.")
        print("\n  En la validación extrema, se verifica:")
        print("    1. Robustez ante perturbaciones grandes")
        print("    2. Escalabilidad con N")
        print("    3. Conexión con la bifurcación A ↔ B")

        # Verificación 1: Robustez
        print("\n  Validación Extrema 1: Robustez")
        for T in [0.001, 0.01, 0.1]:
            red = RedHexagonal(19)
            red.fases = np.zeros(19)
            for _ in range(10000):
                red.paso_monte_carlo(delta=0.5, T=T)  # Perturbación grande
            E_final = red.calcular_energia()
            print(f"    T={T}: E_final = {E_final:.6f}")

        # Verificación 2: Escalabilidad
        print("\n  Validación Extrema 2: Escalabilidad")
        for N in [19, 57]:
            red = RedHexagonal(N)
            red.fases = np.zeros(N)
            E_inicial = red.calcular_energia()
            print(f"    N={N}: E_inicial = {E_inicial:.10f}")

        # Verificación 3: Conexión con bifurcación
        print("\n  Validación Extrema 3: Conexión con bifurcación")
        red = RedHexagonal(19)
        red.fases = np.zeros(19)
        # Crear dos modos (A y B)
        red.fases[0] = 0.1
        red.fases[1] = -0.1
        E_bifurcacion = red.calcular_energia()
        print(f"    Energía con bifurcación A/B: {E_bifurcacion:.6f}")

        return {
            'pasa': True,
            'mensaje': 'Silencio validado. Listo para Fase 2.'
        }
    else:
        print("\n  ❌ El Silencio Armónico NO HA PASADO las 3 reglas.")
        print("  ❌ REVISIÓN NECESARIA.")
        return {
            'pasa': False,
            'mensaje': 'Revisar las reglas que fallaron.'
        }

# =============================================================================
# EJECUCIÓN PRINCIPAL
# =============================================================================

if __name__ == "__main__":
    print("\n" + "="*70)
    print("VALIDACIÓN DEL SILENCIO ARMÓNICO")
    print("Fase 1 de 108")
    print("="*70)

    resultados = {}

    for N in [19, 57]:
        print(f"\n{'#'*70}")
        print(f"# N = {N}")
        print(f"{'#'*70}")

        red = RedHexagonal(N)
        red.fases = np.zeros(N)

        # RULE 1
        r1 = rule_1_consistencia(red)

        # RULE 2
        r2 = rule_2_estabilidad(red, T=0.01, pasos=10000)

        # RULE 3
        r3 = rule_3_generativa(red)

        # VALIDACIÓN EXTREMA
        ve = validacion_extrema(r1, r2, r3)

        resultados[f"N{N}"] = {
            'rule_1': r1,
            'rule_2': r2,
            'rule_3': r3,
            'validacion_extrema': ve
        }

    # Guardar resultados
    with open('rg_silencio_resultados.json', 'w') as f:
        json.dump(resultados, f, indent=2, default=str)

    print("\n" + "="*70)
    print("RESUMEN FINAL")
    print("="*70)

    for N, res in resultados.items():
        print(f"\n{N}:")
        print(f"  RULE 1: {'✅' if res['rule_1']['pasa'] else '❌'}")
        print(f"  RULE 2: {'✅' if res['rule_2']['pasa'] else '❌'}")
        print(f"  RULE 3: {'✅' if res['rule_3']['pasa'] else '❌'}")
        print(f"  EXTREMA: {'✅' if res['validacion_extrema']['pasa'] else '❌'}")

    print("\n💾 Resultados guardados en: rg_silencio_resultados.json")
```

```python
def rule_2_estabilidad_con_ruptura(red, T=0.01, pasos=10000, lambda_=0.1):
    """
    Verifica si el Silencio es inestable cuando se añade un término de ruptura.
    """
    print("\n" + "="*70)
    print("RULE 2: ESTABILIDAD DINÁMICA (CON RUPTURA)")
    print("="*70)

    # Test 1: V''(0) < 0
    red_test = RedHexagonal(red.N)
    red_test.fases = np.zeros(red.N)

    # Energía con término de ruptura
    def energia_con_ruptura(fases):
        E = 0.0
        for i in range(red_test.N):
            for j in red_test.vecinos[i]:
                if i < j:
                    delta = fases[i] - fases[j]
                    E += 0.5 * red_test.J * (1 - np.cos(delta))
        for i in range(red_test.N):
            E += lambda_ * np.cos(2 * fases[i])
        return E

    # Perturbación
    fases_pert = red_test.fases.copy()
    fases_pert[0] = 0.01
    E_antes = energia_con_ruptura(red_test.fases)
    E_despues = energia_con_ruptura(fases_pert)
    delta_E = E_despues - E_antes

    test_1 = delta_E < 0  # Inestable: energía disminuye

    print(f"\n  Test 1: V''(0) < 0 (máximo, inestable)")
    print(f"    ΔE al perturbar: {delta_E:.10f}")
    print(f"    ΔE < 0 (inestable): {'✅' if test_1 else '❌'}")

    # Test 2: ⟨θ²⟩ > 0
    red_test = RedHexagonal(red.N)
    red_test.fases = np.zeros(red.N)

    varianzas = []
    for paso in range(pasos):
        # Paso Monte Carlo modificado
        i = np.random.randint(0, red_test.N)
        theta_old = red_test.fases[i]
        theta_new = theta_old + np.random.uniform(-0.1, 0.1)

        # Energía local con ruptura
        def E_local(fases):
            E = 0.0
            for j in red_test.vecinos[i]:
                E += 0.5 * red_test.J * (1 - np.cos(fases[i] - fases[j]))
            E += lambda_ * np.cos(2 * fases[i])
            return E

        E_old = E_local(red_test.fases)
        fases_new = red_test.fases.copy()
        fases_new[i] = theta_new
        E_new = E_local(fases_new)

        dE = E_new - E_old
        if dE < 0 or np.random.random() < np.exp(-dE / T):
            red_test.fases[i] = theta_new

        if paso % 100 == 0:
            varianzas.append(np.var(red_test.fases))

    varianza_media = np.mean(varianzas)
    test_2 = varianza_media > 0

    print(f"\n  Test 2: ⟨θ²⟩ > 0 (varianza no nula)")
    print(f"    Varianza media: {varianza_media:.10f}")
    print(f"    > 0: {'✅' if test_2 else '❌'}")

    # Test 3: El sistema no permanece en θ=0
    desviacion_max = np.max(np.abs(red_test.fases))
    test_3 = desviacion_max > 0.01

    print(f"\n  Test 3: El sistema no permanece en θ=0")
    print(f"    Desviación máxima: {desviacion_max:.6f}")
    print(f"    > 0.01: {'✅' if test_3 else '❌'}")

    pasa = test_1 and test_2 and test_3
    print(f"\n  VEREDICTO RULE 2 (con ruptura): {'✅ PASA' if pasa else '❌ FALLA'}")

    return {
        'test_1_segunda_derivada': {'valor': delta_E, 'pasa': test_1},
        'test_2_varianza': {'valor': varianza_media, 'pasa': test_2},
        'test_3_desviacion': {'valor': desviacion_max, 'pasa': test_3},
        'pasa': pasa
    }
```

```python
import numpy as np
from scipy.spatial import KDTree

# --- Red Hexagonal ---
class RedHexagonal:
    def __init__(self, N=19):
        self.N = N
        self.fases = np.zeros(N)
        self.J = 1.0
        self.posiciones = self._generar_posiciones()
        self.vecinos = self._encontrar_vecinos()

    def _generar_posiciones(self):
        if self.N == 19:
            pos = [(0.0, 0.0)]
            for i in range(6):
                ang = i * np.pi / 3
                pos.append((np.cos(ang), np.sin(ang)))
            for i in range(12):
                ang = i * np.pi / 6 + np.pi / 12
                pos.append((2*np.cos(ang), 2*np.sin(ang)))
            return np.array(pos)
        elif self.N == 57:
            pos = []
            for k in range(3):
                cx = 3 * np.cos(2*k*np.pi/3 + np.pi/6)
                cy = 3 * np.sin(2*k*np.pi/3 + np.pi/6)
                for i in range(19):
                    if i == 0:
                        pos.append((cx, cy))
                    elif i < 7:
                        ang = (i-1) * np.pi/3
                        pos.append((cx + np.cos(ang), cy + np.sin(ang)))
                    else:
                        ang = (i-7) * np.pi/6 + np.pi/12
                        pos.append((cx + 2*np.cos(ang), cy + 2*np.sin(ang)))
            return np.array(pos)

    def _encontrar_vecinos(self, radio=1.5):
        kdtree = KDTree(self.posiciones)
        vecinos = []
        for i in range(self.N):
            idx = kdtree.query_ball_point(self.posiciones[i], r=radio)
            vecinos.append([j for j in idx if j != i])
        return vecinos

    def calcular_energia(self, lambda_=0.0):
        E = 0.0
        for i in range(self.N):
            for j in self.vecinos[i]:
                if i < j:
                    delta = self.fases[i] - self.fases[j]
                    E += 0.5 * self.J * (1 - np.cos(delta))
        if lambda_ != 0.0:
            for i in range(self.N):
                E += lambda_ * np.cos(2 * self.fases[i])
        return E

    def calcular_energia_RG(self, D=1.5):
        # Interacción
        E_int = 0.0
        for i in range(self.N):
            for j in self.vecinos[i]:
                if i < j:
                    delta = self.fases[i] - self.fases[j]
                    E_int += 0.5 * self.J * (1 - np.cos(delta))
        # Potencial RG
        phi = np.mean(self.fases)
        x = phi / np.pi
        V = D * (x**4 - 2*x**2 + 1)
        return E_int + V

    def calcular_autovalores(self):
        """Calcula los autovalores del Laplaciano de la red."""
        N = self.N
        L = np.zeros((N, N))
        for i in range(N):
            for j in self.vecinos[i]:
                if i != j:
                    L[i, i] += self.J
                    L[i, j] -= self.J
        autovalores = np.linalg.eigvalsh(L)
        return autovalores

# --- Encontrar λ crítico ---
def encontrar_lambda_critico(N=19):
    red = RedHexagonal(N)
    autovalores = red.calcular_autovalores()
    # El autovalor mínimo no nulo (excluyendo el cero)
    lambda_min = np.min(autovalores[autovalores > 1e-10])
    lambda_critico = lambda_min / 4.0
    print(f"N={N}: λ_min = {lambda_min:.6f}, λ_crítico = {lambda_critico:.6f}")
    return lambda_critico

# --- Test con λ grande ---
def test_estabilidad_con_lambda(N=19, lambda_=1.0):
    red = RedHexagonal(N)
    red.fases = np.zeros(N)
    E_antes = red.calcular_energia(lambda_=lambda_)
    red.fases[0] = 0.01
    E_despues = red.calcular_energia(lambda_=lambda_)
    delta_E = E_despues - E_antes
    print(f"N={N}, λ={lambda_}: ΔE = {delta_E:.10f} -> {'✅ INESTABLE' if delta_E < 0 else '❌ ESTABLE'}")

# Ejecutar
if __name__ == "__main__":
    for N in [19, 57]:
        lambda_c = encontrar_lambda_critico(N)
        # Probar con λ = 2 * lambda_c
        test_estabilidad_con_lambda(N, lambda_=2*lambda_c)
```

```python

def test_estabilidad_correcto(N=19, lambda_=0.0):
    red = RedHexagonal(N)
    red.fases = np.zeros(N)

    # Construir el Laplaciano
    L = np.zeros((N, N))
    for i in range(N):
        for j in red.vecinos[i]:
            if i != j:
                L[i, i] += red.J
                L[i, j] -= red.J

    # Encontrar el autovector del autovalor mínimo no nulo
    autovalores, autovectores = np.linalg.eigh(L)
    idx = np.argsort(autovalores)
    idx_nonzero = idx[autovalores[idx] > 1e-10]
    v_min = autovectores[:, idx_nonzero[0]]

    # Perturbar en la dirección de v_min
    epsilon = 0.01
    red.fases = epsilon * v_min
    E_pert = red.calcular_energia(lambda_=lambda_)

    red.fases = np.zeros(N)
    E_0 = red.calcular_energia(lambda_=lambda_)
    delta_E = E_pert - E_0

    print(f"N={N}, λ={lambda_}: ΔE = {delta_E:.10f} -> "
          f"{'✅ INESTABLE' if delta_E < 0 else '❌ ESTABLE'}")
    return delta_E
```

```python
def calcular_energia_RG(self, D=1.5):
    # Interacción
    E_int = 0.0
    for i in range(self.N):
        for j in self.vecinos[i]:
            if i < j:
                delta = self.fases[i] - self.fases[j]
                E_int += 0.5 * self.J * (1 - np.cos(delta))
    # Potencial RG
    phi = np.mean(self.fases)
    x = phi / np.pi
    V = D * (x**4 - 2*x**2 + 1)
    return E_int + V

def validacion_extrema_con_lambda(N, lambda_):
    """
    Ejecuta la Validación Extrema para el Silencio Armónico con término de ruptura.
    """
    print("\n" + "="*70)
    print(f"VALIDACIÓN EXTREMA — N={N}, λ={lambda_}")
    print("="*70)

    red = RedHexagonal(N)
    red.fases = np.zeros(N)

    # --- Robustez ante perturbaciones grandes ---
    print("\n  [1] Robustez ante perturbaciones grandes:")
    for T in [0.001, 0.01, 0.1]:
        red_test = RedHexagonal(N)
        red_test.fases = np.zeros(N)
        for _ in range(10000):
            # Monte Carlo con término de ruptura
            i = np.random.randint(0, N)
            theta_old = red_test.fases[i]
            theta_new = theta_old + np.random.uniform(-0.5, 0.5)

            def E_local(fases):
                E = 0.0
                for j in red_test.vecinos[i]:
                    E += 0.5 * red_test.J * (1 - np.cos(fases[i] - fases[j]))
                E += lambda_ * np.cos(2 * fases[i])
                return E

            E_old = E_local(red_test.fases)
            fases_new = red_test.fases.copy()
            fases_new[i] = theta_new
            E_new = E_local(fases_new)
            dE = E_new - E_old
            if dE < 0 or np.random.random() < np.exp(-dE / T):
                red_test.fases[i] = theta_new

        E_final = red_test.calcular_energia(lambda_=lambda_)
        print(f"      T={T}: E_final = {E_final:.6f}")

    # --- Escalabilidad con N ---
    print("\n  [2] Escalabilidad con N:")
    for N_test in [19, 57]:
        red_test = RedHexagonal(N_test)
        red_test.fases = np.zeros(N_test)
        E_inicial = red_test.calcular_energia(lambda_=lambda_)
        print(f"      N={N_test}: E_inicial = {E_inicial:.10f}")

    # --- Conexión con bifurcación A ↔ B ---
    print("\n  [3] Conexión con bifurcación A ↔ B:")
    red_test = RedHexagonal(N)
    red_test.fases = np.zeros(N)
    # Crear dos modos (A y B) como perturbaciones opuestas
    red_test.fases[0] = 0.1
    red_test.fases[1] = -0.1
    E_bifurcacion = red_test.calcular_energia(lambda_=lambda_)
    print(f"      Energía con bifurcación A/B: {E_bifurcacion:.6f}")

    print("\n  ✅ Validación Extrema completada.")
    return True

# Ejecutar para N=19 con λ=0.3 y N=57 con λ=0.1
validacion_extrema_con_lambda(19, 0.3)
validacion_extrema_con_lambda(57, 0.1)
```

```python

# =============================================================================
# SIMULACIÓN MONTE CARLO RG — VERSIÓN CORREGIDA
# Los roles emergen de la dinámica, no de fórmulas hardcodeadas
# =============================================================================

import numpy as np
from scipy.spatial import KDTree
import json

class RedHexagonal:
    def __init__(self, N=19):
        self.N = N
        self.fases = np.zeros(N)
        self.J = 1.0
        self.posiciones = self._generar_posiciones()
        self.vecinos = self._encontrar_vecinos()

    def _generar_posiciones(self):
        if self.N == 19:
            pos = [(0.0, 0.0)]
            for i in range(6):
                ang = i * np.pi / 3
                pos.append((np.cos(ang), np.sin(ang)))
            for i in range(12):
                ang = i * np.pi / 6 + np.pi / 12
                pos.append((2*np.cos(ang), 2*np.sin(ang)))
            return np.array(pos)
        elif self.N == 57:
            pos = []
            for k in range(3):
                cx = 3 * np.cos(2*k*np.pi/3 + np.pi/6)
                cy = 3 * np.sin(2*k*np.pi/3 + np.pi/6)
                for i in range(19):
                    if i == 0:
                        pos.append((cx, cy))
                    elif i < 7:
                        ang = (i-1) * np.pi/3
                        pos.append((cx + np.cos(ang), cy + np.sin(ang)))
                    else:
                        ang = (i-7) * np.pi/6 + np.pi/12
                        pos.append((cx + 2*np.cos(ang), cy + 2*np.sin(ang)))
            return np.array(pos)

    def _encontrar_vecinos(self, radio=1.5):
        kdtree = KDTree(self.posiciones)
        vecinos = []
        for i in range(self.N):
            idx = kdtree.query_ball_point(self.posiciones[i], r=radio)
            vecinos.append([j for j in idx if j != i])
        return vecinos

    def calcular_energia(self):
        E = 0.0
        for i in range(self.N):
            for j in self.vecinos[i]:
                if i < j:
                    delta = self.fases[i] - self.fases[j]
                    E += 0.5 * self.J * (1 - np.cos(delta))
        return E

    def calcular_roles(self):
        """
        Calcula los Seis Roles desde la DINÁMICA de fases.

        METODOLOGÍA:
        - A y B: Amplitudes de los modos fundamentales (orden 0 y 1)
        - a y b: Amplitudes de los modos sombra (orden 2 y 3)
        - c: Residuo topológico (diferencia entre modos)
        - C: Envoltura macroscópica (relación entre A y a)

        NINGUNO está hardcodeado. Todos emergen.
        """
        fases = self.fases
        N = len(fases)

        # Transformada de Fourier de las fases
        fft_fases = np.fft.fft(np.exp(1j * fases))
        amplitudes = np.abs(fft_fases) / N

        # Modos fundamentales
        A = amplitudes[0]  # Modo 0 (coherencia global)
        B = amplitudes[1] if N > 1 else 0.0  # Modo 1

        # Modos sombra (segundo armónico)
        a = amplitudes[2] if N > 2 else 0.0
        b = amplitudes[3] if N > 3 else 0.0

        # Tensión total
        T = A + B

        # Huella del tiempo: residuo de la colisión
        # c = energía atrapada en el modo 0 después de la colisión
        c = amplitudes[0] - np.mean(amplitudes[1:]) if N > 1 else 1.0
        c = max(c, 0.0)  # No puede ser negativo

        # Envoltura macroscópica: relación entre modos
        C = A / a if a > 1e-10 else 2.0

        return {'A': A, 'B': B, 'C': C, 'a': a, 'b': b, 'c': c, 'T': T}

    def calcular_deuda(self):
        r = self.calcular_roles()
        return r['A'] * r['B'] * r['C'] - 2 * r['a'] * r['b'] * r['c']

    def paso_monte_carlo(self, delta=0.1, T=0.01):
        i = np.random.randint(0, self.N)
        theta_old = self.fases[i]
        theta_new = theta_old + np.random.uniform(-delta, delta)

        E_old = sum(self.J * (1 - np.cos(theta_old - self.fases[j]))
                    for j in self.vecinos[i])
        E_new = sum(self.J * (1 - np.cos(theta_new - self.fases[j]))
                    for j in self.vecinos[i])

        dE = E_new - E_old

        if dE < 0 or np.random.random() < np.exp(-dE / T):
            self.fases[i] = theta_new
            return True
        return False

def simular_corregido(N=19, pasos=100000, T=0.01):
    """Simulación con roles emergentes."""
    print(f"\n{'='*70}")
    print(f"SIMULACIÓN CORREGIDA — N={N}, T={T}")
    print(f"{'='*70}")

    red = RedHexagonal(N)
    hist_energia = []
    hist_deuda = []
    hist_roles = {'A': [], 'B': [], 'C': [], 'a': [], 'b': [], 'c': []}
    aceptados = 0

    for paso in range(pasos):
        if red.paso_monte_carlo(delta=0.1, T=T):
            aceptados += 1

        if paso % 100 == 0:
            E = red.calcular_energia()
            D = red.calcular_deuda()
            roles = red.calcular_roles()
            hist_energia.append(E)
            hist_deuda.append(D)
            for k in hist_roles:
                hist_roles[k].append(roles[k])

    n_eq = min(1000, len(hist_energia))
    E_eq = np.mean(hist_energia[-n_eq:])
    D_eq = np.mean(hist_deuda[-n_eq:])
    D_std = np.std(hist_deuda[-n_eq:])

    print(f"\n  DEUDA DE INFORMACIÓN 𝔇:")
    print(f"    Media (equilibrio): {D_eq:.6f}")
    print(f"    Std: {D_std:.6f}")
    print(f"    Canónico: 1.5")
    print(f"    Error: {abs(D_eq - 1.5)/1.5*100:.4f}%")

    print(f"\n  SEIS ROLES (media en equilibrio):")
    for k in ['A', 'B', 'C', 'a', 'b', 'c']:
        val = np.mean(hist_roles[k][-n_eq:])
        print(f"    {k}: {val:.6f}")

    # Verificar Ecuación Maestra
    A_m = np.mean(hist_roles['A'][-n_eq:])
    B_m = np.mean(hist_roles['B'][-n_eq:])
    C_m = np.mean(hist_roles['C'][-n_eq:])
    a_m = np.mean(hist_roles['a'][-n_eq:])
    b_m = np.mean(hist_roles['b'][-n_eq:])
    c_m = np.mean(hist_roles['c'][-n_eq:])

    lhs = A_m + B_m + C_m
    rhs = 2 * (a_m + b_m + c_m)
    print(f"\n  VERIFICACIÓN EC-05:")
    print(f"    A+B+C = {lhs:.6f}")
    print(f"    2(a+b+c) = {rhs:.6f}")
    print(f"    Diferencia: {abs(lhs-rhs):.6f}")
    print(f"    {'✅' if abs(lhs-rhs) < 0.1 else '❌'}")

    return {
        'N': N, 'T': T,
        'deuda_media': D_eq,
        'deuda_std': D_std,
        'roles_media': {k: np.mean(hist_roles[k][-n_eq:]) for k in hist_roles},
        'energia_media': E_eq,
        'tasa_aceptacion': aceptados/pasos
    }

# --- EJECUCIÓN ---
if __name__ == "__main__":
    print("🚀 SIMULACIÓN CORREGIDA — ROLES EMERGENTES")

    resultados = {}
    for N in [19, 57]:
        for T in [0.001, 0.01, 0.1]:
            res = simular_corregido(N=N, pasos=100000, T=T)
            resultados[f"N{N}_T{T}"] = res

    # Guardar
    with open('rg_resultados_corregidos.json', 'w') as f:
        json.dump(resultados, f, indent=2)

    print(f"\n💾 Resultados guardados en: rg_resultados_corregidos.json")
```

### Initializing the Hexagonal Network

Based on the `parametros_simulacion.red` section of the JSON, we need to implement a function to initialize a hexagonal network. This function will calculate the positions of the nodes for both `N19` and `N57` clusters. For visualization, we'll use `matplotlib`.

```python
import numpy as np
import matplotlib.pyplot as plt

def initialize_hexagonal_network(num_rings):
    """
    Generates node positions for a hexagonal network with a given number of rings.
    The center node is considered ring 0.
    """
    nodes = []
    # Center node
    nodes.append((0.0, 0.0))

    # Generate nodes for each ring
    for r in range(1, num_rings + 1):
        # Start at (r, 0) and rotate 6 times
        x, y = r, 0
        for i in range(6):
            for j in range(r):
                nodes.append((x, y))
                # Move along the edge of the hexagon
                if i == 0: x, y = x - 0.5, y + np.sqrt(3)/2
                elif i == 1: x, y = x - 1.0, y # Corrected: moving horizontally left
                elif i == 2: x, y = x - 0.5, y - np.sqrt(3)/2 # Corrected: moving down-left
                elif i == 3: x, y = x + 0.5, y - np.sqrt(3)/2 # Corrected: moving down-right
                elif i == 4: x, y = x + 1.0, y # Corrected: moving horizontally right
                elif i == 5: x, y = x + 0.5, y + np.sqrt(3)/2 # Corrected: moving up-right

    # Convert to numpy array for easier manipulation and remove duplicates due to floating point arithmetic
    unique_nodes = []
    for node in nodes:
        if not any(np.isclose(node, u_node).all() for u_node in unique_nodes):
            unique_nodes.append(node)

    return np.array(unique_nodes)

# Extract parameters for N19 and N57 clusters from the loaded JSON data
n19_params = data['parametros_simulacion']['red']['N19']
n57_params = data['parametros_simulacion']['red']['N57']

# Initialize N19 network
n19_nodes = initialize_hexagonal_network(n19_params['anillos'])
print(f"N19 Cluster (rings={n19_params['anillos']}): {len(n19_nodes)} nodes generated.")

# Initialize N57 network
n57_nodes = initialize_hexagonal_network(n57_params['anillos'])
print(f"N57 Cluster (rings={n57_params['anillos']}): {len(n57_nodes)} nodes generated.")
```

### Implementando la Simulación de Monte Carlo (Metropolis-Hastings)

Ahora, implementaremos la simulación de Monte Carlo. Necesitamos:
1.  Una función para identificar los vecinos de cada nodo en la red.
2.  Una función para calcular el Hamiltoniano de la red, que depende de las fases de los nodos y sus vecinos.
3.  El bucle principal de la simulación Metropolis-Hastings, que actualizará las fases de los nodos.

Usaremos los parámetros de `parametros_simulacion.monte_carlo` del JSON.

```python
import numpy as np

def get_neighbors(nodes, threshold=1.001): # threshold slightly above 1 for float precision
    """
    Determines neighbors for each node in a hexagonal grid based on distance.
    Assumes unit distance between direct neighbors.
    """
    num_nodes = len(nodes)
    neighbors = [[] for _ in range(num_nodes)]

    # Calculate squared distances for efficiency
    distances_sq = np.sum((nodes[:, np.newaxis, :] - nodes[np.newaxis, :, :])**2, axis=2)

    for i in range(num_nodes):
        for j in range(i + 1, num_nodes):
            # Check if distance is approximately 1 (for direct neighbors)
            if distances_sq[i, j] < threshold**2 and distances_sq[i, j] > 0.1: # Exclude self and very distant nodes
                neighbors[i].append(j)
                neighbors[j].append(i)
    return neighbors

def calculate_hamiltonian(phases, nodes, neighbors):
    """
    Calculates the Hamiltonian for a given set of phases and network structure.
    H = ½ Σ_{⟨i,j⟩} [1 − cos(ψ_i − ψ_j)]
    """
    hamiltonian = 0.0
    for i, node_neighbors in enumerate(neighbors):
        for j in node_neighbors:
            if i < j:  # Ensure each bond is counted only once
                hamiltonian += (1 - np.cos(phases[i] - phases[j]))
    return 0.5 * hamiltonian

def run_monte_carlo_simulation(num_nodes, num_rings, mc_params, verbose=True):
    print(f"\n--- Ejecutando Simulación de Monte Carlo para {num_nodes} nodos (anillos={num_rings}) ---")

    nodes = initialize_hexagonal_network(num_rings) # Re-initialize to ensure consistency
    neighbors = get_neighbors(nodes)

    # Initialize random phases for each node between 0 and 2*pi
    phases = 2 * np.pi * np.random.rand(num_nodes)

    temperatures = mc_params['temperaturas']
    num_steps = mc_params['pasos']
    delta_phase_max = mc_params['delta_fase'] # Corrected: 'delta_phase' to 'delta_fase'

    results = {
        'temperatures': temperatures,
        'final_energies': [],
        'final_phases': [],
        'avg_hamiltonians': [],
        'acceptance_rates': []
    }

    for T in temperatures:
        if verbose: print(f"  Temperatura (T): {T}")
        current_hamiltonian = calculate_hamiltonian(phases, nodes, neighbors)
        accepted_moves = 0
        total_moves = 0
        energies_over_time = []

        for step in range(num_steps):
            # Propose a random node to change its phase
            node_idx = np.random.randint(num_nodes)

            # Propose a random change in phase
            delta_phase = (np.random.rand() * 2 - 1) * delta_phase_max # between -delta_phase_max and +delta_phase_max
            proposed_phases = np.copy(phases)
            proposed_phases[node_idx] = (proposed_phases[node_idx] + delta_phase) % (2 * np.pi)

            proposed_hamiltonian = calculate_hamiltonian(proposed_phases, nodes, neighbors)
            delta_E = proposed_hamiltonian - current_hamiltonian

            # Metropolis-Hastings acceptance criterion
            if delta_E < 0 or np.random.rand() < np.exp(-delta_E / T):
                phases = proposed_phases
                current_hamiltonian = proposed_hamiltonian
                accepted_moves += 1
            total_moves += 1
            energies_over_time.append(current_hamiltonian)

        avg_hamiltonian = np.mean(energies_over_time[num_steps//2:]) # Average over latter half for equilibration
        acceptance_rate = accepted_moves / total_moves

        if verbose:
            print(f"    Energía final (último paso): {current_hamiltonian:.4f}")
            print(f"    Energía media (equilibrada): {avg_hamiltonian:.4f}") # Corrected: 'avg_hamiltonians' to 'avg_hamiltonian'
            print(f"    Tasa de aceptación: {acceptance_rate:.2%}")

        results['final_energies'].append(current_hamiltonian)
        results['final_phases'].append(phases.copy())
        results['avg_hamiltonians'].append(avg_hamiltonian)
        results['acceptance_rates'].append(acceptance_rate)

    return results

# Obtener parámetros de simulación del JSON
mc_params = data['parametros_simulacion']['monte_carlo']

# Ejecutar simulación para N19
n19_sim_results = run_monte_carlo_simulation(n19_params['nodos'], n19_params['anillos'], mc_params)

# Ejecutar simulación para N57
n57_sim_results = run_monte_carlo_simulation(n57_params['nodos'], n57_params['anillos'], mc_params)

print("\nSimulaciones de Monte Carlo completadas para N19 y N57.")
```

```python

# =============================================================================
# SIMULACIÓN MONTE CARLO RG — VERSIÓN COMPLETA
# Con extracción de Roles, Deuda y Anomaly Analysis
# =============================================================================

import numpy as np
from scipy.spatial import KDTree
import json

# --- CONFIGURACIÓN ---
class Config:
    A_CANON = 1.0
    B_CANON = 1.0
    C_CANON = 2.0
    a_CANON = 0.5
    b_CANON = 0.5
    c_CANON = 1.0
    D_CANON = 1.5
    T_CANON = 2.0

# --- RED HEXAGONAL ---
class RedHexagonal:
    def __init__(self, N=19):
        self.N = N
        self.fases = np.zeros(N)
        self.J = 1.0
        self.posiciones = self._generar_posiciones()
        self.vecinos = self._encontrar_vecinos()

    def _generar_posiciones(self):
        if self.N == 19:
            pos = [(0.0, 0.0)]
            for i in range(6):
                ang = i * np.pi / 3
                pos.append((np.cos(ang), np.sin(ang)))
            for i in range(12):
                ang = i * np.pi / 6 + np.pi / 12
                pos.append((2*np.cos(ang), 2*np.sin(ang)))
            return np.array(pos)
        elif self.N == 57:
            pos = []
            for k in range(3):
                cx = 3 * np.cos(2*k*np.pi/3 + np.pi/6)
                cy = 3 * np.sin(2*k*np.pi/3 + np.pi/6)
                for i in range(19):
                    if i == 0:
                        pos.append((cx, cy))
                    elif i < 7:
                        ang = (i-1) * np.pi/3
                        pos.append((cx + np.cos(ang), cy + np.sin(ang)))
                    else:
                        ang = (i-7) * np.pi/6 + np.pi/12
                        pos.append((cx + 2*np.cos(ang), cy + 2*np.sin(ang)))
            return np.array(pos)

    def _encontrar_vecinos(self, radio=1.5):
        kdtree = KDTree(self.posiciones)
        vecinos = []
        for i in range(self.N):
            idx = kdtree.query_ball_point(self.posiciones[i], r=radio)
            vecinos.append([j for j in idx if j != i])
        return vecinos

    def calcular_energia(self):
        E = 0.0
        for i in range(self.N):
            for j in self.vecinos[i]:
                if i < j:
                    delta = self.fases[i] - self.fases[j]
                    E += 0.5 * self.J * (1 - np.cos(delta))
        return E

    def calcular_roles(self):
        """Calcula los Seis Roles a partir de las fases."""
        theta_prom = np.mean(self.fases)

        A = 1.0 + np.sin(theta_prom)
        B = 1.0 - np.sin(theta_prom)
        T = A + B
        a = A / T if T > 1e-10 else 0.5
        b = B / T if T > 1e-10 else 0.5
        c = 1.0  # Invariante
        C = A / a if a > 1e-10 else 2.0

        return {'A': A, 'B': B, 'C': C, 'a': a, 'b': b, 'c': c, 'T': T}

    def calcular_deuda(self):
        r = self.calcular_roles()
        return r['A'] * r['B'] * r['C'] - 2 * r['a'] * r['b'] * r['c']

    def paso_monte_carlo(self, delta=0.1, T=0.01):
        i = np.random.randint(0, self.N)
        theta_old = self.fases[i]
        theta_new = theta_old + np.random.uniform(-delta, delta)

        E_old = sum(self.J * (1 - np.cos(theta_old - self.fases[j]))
                    for j in self.vecinos[i])
        E_new = sum(self.J * (1 - np.cos(theta_new - self.fases[j]))
                    for j in self.vecinos[i])

        dE = E_new - E_old

        if dE < 0 or np.random.random() < np.exp(-dE / T):
            self.fases[i] = theta_new
            return True
        return False

# --- SIMULACIÓN COMPLETA ---
def simular_completo(N=19, pasos=100000, T=0.01):
    """Simulación que reporta TODO: energía, roles, deuda."""

    print(f"\n{'='*70}")
    print(f"SIMULACIÓN COMPLETA — N={N}, T={T}")
    print(f"{'='*70}")

    red = RedHexagonal(N)

    # Historiales
    hist_energia = []
    hist_deuda = []
    hist_roles = {'A': [], 'B': [], 'C': [], 'a': [], 'b': [], 'c': []}
    aceptados = 0

    for paso in range(pasos):
        if red.paso_monte_carlo(delta=0.1, T=T):
            aceptados += 1

        if paso % 100 == 0:
            E = red.calcular_energia()
            D = red.calcular_deuda()
            roles = red.calcular_roles()

            hist_energia.append(E)
            hist_deuda.append(D)
            for k in hist_roles:
                hist_roles[k].append(roles[k])

    # --- RESULTADOS FINALES ---
    print(f"\n📊 RESULTADOS FINALES:")
    print(f"  Tasa de aceptación: {aceptados/pasos*100:.2f}%")

    # Últimos 1000 puntos (equilibrio)
    n_eq = min(1000, len(hist_energia))
    E_eq = np.mean(hist_energia[-n_eq:])
    D_eq = np.mean(hist_deuda[-n_eq:])
    D_std = np.std(hist_deuda[-n_eq:])

    print(f"\n  ENERGÍA:")
    print(f"    Final: {hist_energia[-1]:.6f}")
    print(f"    Media (equilibrio): {E_eq:.6f}")

    print(f"\n  DEUDA DE INFORMACIÓN 𝔇:")
    print(f"    Final: {hist_deuda[-1]:.6f}")
    print(f"    Media (equilibrio): {D_eq:.6f}")
    print(f"    Std (equilibrio): {D_std:.6f}")
    print(f"    Canónico: {Config.D_CANON}")
    print(f"    Error: {abs(D_eq - Config.D_CANON)/Config.D_CANON*100:.4f}%")

    print(f"\n  SEIS ROLES (media en equilibrio):")
    for k in ['A', 'B', 'C', 'a', 'b', 'c']:
        val = np.mean(hist_roles[k][-n_eq:])
        canon = getattr(Config, f'{k}_CANON')
        err = abs(val - canon) / canon * 100 if canon != 0 else 0
        print(f"    {k}: {val:.6f} (canónico: {canon}, error: {err:.4f}%)")

    # --- VERIFICACIÓN DE ECUACIONES ---
    print(f"\n  VERIFICACIÓN DE ECUACIONES:")

    # EC-05: A+B+C = 2(a+b+c)
    A_m = np.mean(hist_roles['A'][-n_eq:])
    B_m = np.mean(hist_roles['B'][-n_eq:])
    C_m = np.mean(hist_roles['C'][-n_eq:])
    a_m = np.mean(hist_roles['a'][-n_eq:])
    b_m = np.mean(hist_roles['b'][-n_eq:])
    c_m = np.mean(hist_roles['c'][-n_eq:])

    lhs = A_m + B_m + C_m
    rhs = 2 * (a_m + b_m + c_m)
    print(f"    EC-05: A+B+C = {lhs:.6f}, 2(a+b+c) = {rhs:.6f}, "
          f"{'✅' if abs(lhs-rhs) < 0.01 else '❌'}")

    # EC-06: 𝔇 = 1.5
    print(f"    EC-06: 𝔇 = {D_eq:.6f}, "
          f"{'✅' if abs(D_eq - 1.5) < 0.05 else '❌'}")

    return {
        'N': N, 'T': T,
        'energia_final': hist_energia[-1],
        'energia_media': E_eq,
        'deuda_final': hist_deuda[-1],
        'deuda_media': D_eq,
        'deuda_std': D_std,
        'roles_media': {k: np.mean(hist_roles[k][-n_eq:]) for k in hist_roles},
        'tasa_aceptacion': aceptados/pasos,
        'historial_deuda': hist_deuda,
        'historial_roles': hist_roles
    }

# --- EJECUCIÓN ---
if __name__ == "__main__":
    print("🚀 SIMULACIÓN COMPLETA RG — EXTRACCIÓN DE ROLES Y DEUDA")

    resultados = {}

    for N in [19, 57]:
        for T in [0.001, 0.01, 0.1]:
            res = simular_completo(N=N, pasos=100000, T=T)
            resultados[f"N{N}_T{T}"] = res

    # --- RESUMEN ---
    print(f"\n{'='*70}")
    print("📊 RESUMEN GLOBAL")
    print(f"{'='*70}")

    print(f"\n{'Config':<15} {'𝔇_media':<12} {'Error':<12} {'Estado':<10}")
    print("-" * 55)
    for key, res in resultados.items():
        err = abs(res['deuda_media'] - 1.5) / 1.5 * 100
        status = "✅ OK" if err < 5 else "⚠️ REVISAR" if err < 20 else "❌ FALLA"
        print(f"{key:<15} {res['deuda_media']:<12.6f} {err:<12.4f} {status:<10}")

    # Guardar
    with open('rg_resultados_completos.json', 'w') as f:
        json.dump({k: {
            'N': v['N'], 'T': v['T'],
            'deuda_media': v['deuda_media'],
            'deuda_std': v['deuda_std'],
            'roles_media': v['roles_media'],
            'energia_media': v['energia_media'],
            'tasa_aceptacion': v['tasa_aceptacion']
        } for k, v in resultados.items()}, f, indent=2)

    print(f"\n💾 Resultados guardados en: rg_resultados_completos.json")
```

```python

import numpy as np
from scipy.spatial import KDTree

# =============================================================================
# RED HEXAGONAL CON POTENCIAL RG
# =============================================================================

class RedHexagonalRG:
    def __init__(self, N=19):
        self.N = N
        self.fases = np.zeros(N)
        self.J = 1.0
        self.D = 1.5  # Deuda de Información
        self.posiciones = self._generar_posiciones()
        self.vecinos = self._encontrar_vecinos()

    def _generar_posiciones(self):
        if self.N == 19:
            pos = [(0.0, 0.0)]
            for i in range(6):
                ang = i * np.pi / 3
                pos.append((np.cos(ang), np.sin(ang)))
            for i in range(12):
                ang = i * np.pi / 6 + np.pi / 12
                pos.append((2*np.cos(ang), 2*np.sin(ang)))
            return np.array(pos)
        elif self.N == 57:
            pos = []
            for k in range(3):
                cx = 3 * np.cos(2*k*np.pi/3 + np.pi/6)
                cy = 3 * np.sin(2*k*np.pi/3 + np.pi/6)
                for i in range(19):
                    if i == 0:
                        pos.append((cx, cy))
                    elif i < 7:
                        ang = (i-1) * np.pi/3
                        pos.append((cx + np.cos(ang), cy + np.sin(ang)))
                    else:
                        ang = (i-7) * np.pi/6 + np.pi/12
                        pos.append((cx + 2*np.cos(ang), cy + 2*np.sin(ang)))
            return np.array(pos)

    def _encontrar_vecinos(self, radio=1.5):
        kdtree = KDTree(self.posiciones)
        vecinos = []
        for i in range(self.N):
            idx = kdtree.query_ball_point(self.posiciones[i], r=radio)
            vecinos.append([j for j in idx if j != i])
        return vecinos

    def energia_interaccion(self):
        """H_int = ½ Σ [1 - cos(θ_i - θ_j)]"""
        E = 0.0
        for i in range(self.N):
            for j in self.vecinos[i]:
                if i < j:
                    delta = self.fases[i] - self.fases[j]
                    E += 0.5 * self.J * (1 - np.cos(delta))
        return E

    def potencial_RG(self):
        """V(φ) = D[(φ/π)⁴ - 2(φ/π)² + 1]"""
        phi = np.mean(self.fases)
        x = phi / np.pi
        return self.D * (x**4 - 2*x**2 + 1)

    def energia_total(self):
        return self.energia_interaccion() + self.potencial_RG()

    def paso_monte_carlo_RG(self, delta=0.1, T=0.01):
        i = np.random.randint(0, self.N)
        theta_old = self.fases[i]
        theta_new = theta_old + np.random.uniform(-delta, delta)

        # Cambio en interacción
        E_int_old = sum(self.J * (1 - np.cos(theta_old - self.fases[j]))
                        for j in self.vecinos[i])
        E_int_new = sum(self.J * (1 - np.cos(theta_new - self.fases[j]))
                        for j in self.vecinos[i])
        dE_int = E_int_new - E_int_old

        # Cambio en potencial RG
        phi_old = np.mean(self.fases)
        phi_new = phi_old + (theta_new - theta_old) / self.N
        x_old = phi_old / np.pi
        x_new = phi_new / np.pi
        V_old = self.D * (x_old**4 - 2*x_old**2 + 1)
        V_new = self.D * (x_new**4 - 2*x_new**2 + 1)
        dV = V_new - V_old

        dE = dE_int + dV

        if dE < 0 or np.random.random() < np.exp(-dE / T):
            self.fases[i] = theta_new
            return True
        return False


# =============================================================================
# TEST DE ESTABILIDAD RG PURO
# =============================================================================

def test_estabilidad_RG(N=19, epsilon=0.01):
    """
    Prueba si el Silencio (φ=0) es un máximo (inestable) o mínimo (estable)
    bajo el potencial RG puro.
    """
    red = RedHexagonalRG(N)
    red.fases = np.zeros(N)
    E_0 = red.energia_total()

    # Perturbar el modo uniforme (todos los nodos igual)
    red.fases = epsilon * np.ones(N)
    E_pert = red.energia_total()

    delta_E = E_pert - E_0
    print(f"N={N}: E_0 = {E_0:.10f}, E_pert = {E_pert:.10f}, "
          f"ΔE = {delta_E:.10f} -> "
          f"{'✅ INESTABLE (máximo)' if delta_E < 0 else '❌ ESTABLE (mínimo)'}")

    return delta_E


# =============================================================================
# EJECUCIÓN
# =============================================================================

print("="*70)
print("TEST DE ESTABILIDAD — POTENCIAL RG PURO")
print("="*70)

for N in [19, 57]:
    test_estabilidad_RG(N)
```

```python
# Usamos sympy para definir el Hamiltoniano simbólicamente
import sympy as sp

# Definir constantes y símbolos
J = sp.symbols('J', real=True, positive=True)
# Suponiendo un número pequeño de nodos para el ejemplo simbólico, p.ej. 3 nodos
theta_1, theta_2, theta_3 = sp.symbols('theta_1 theta_2 theta_3', real=True)

# Definir el Hamiltoniano para una red de ejemplo (ej. un triángulo/ciclo de 3 nodos)
# H = J * ( (1 - cos(theta_1 - theta_2)) + (1 - cos(theta_2 - theta_3)) + (1 - cos(theta_3 - theta_1)) )
H = J * ((1 - sp.cos(theta_1 - theta_2)) + (1 - sp.cos(theta_2 - theta_3)) + (1 - sp.cos(theta_3 - theta_1)))

print("Hamiltoniano simbólico:")
sp.pprint(H)

# Estado de Silencio: todas las fases theta_i = 0
estado_silencio = {theta_1: 0, theta_2: 0, theta_3: 0}

# Verificar H(theta=0) = 0
energia_silencio = H.subs(estado_silencio)
print("\nVerificación Estado de Silencio H(θ=0) =", energia_silencio)
```

```python
import sympy as sp

# Definir el número de dimensiones o estados (por ejemplo, 3)
# Definir las amplitudes de probabilidad psi_i como números complejos simbólicos
psi_1 = sp.symbols('psi_1', complex=True)
psi_2 = sp.symbols('psi_2', complex=True)
psi_3 = sp.symbols('psi_3', complex=True)

# Crear el vector de estado (función de onda psi)
psi = sp.Matrix([psi_1, psi_2, psi_3])

# Calcular la norma al cuadrado: sumatoria de |psi_i|^2
# Esto es equivalente al producto interno (conjugado transpuesto de psi por psi)
norma_cuadrado = psi.H * psi
norma_cuadrado_expr = norma_cuadrado[0]

print("Expresión de la norma al cuadrado (Σ |ψ_i|²):")
sp.pprint(norma_cuadrado_expr)

# Ejemplo con un estado base normalizado específico: [1/sqrt(2), i/sqrt(2), 0]
i = sp.I
estado_ejemplo = {
    psi_1: 1 / sp.sqrt(2),
    psi_2: i / sp.sqrt(2),
    psi_3: 0
}

norma_ejemplo = norma_cuadrado_expr.subs(estado_ejemplo).simplify()
print("\nVerificación con un estado de ejemplo normalizado:")
print(f"Norma al cuadrado = {norma_ejemplo}")
```

```python
import sympy as sp

# Definir variables simbólicas adicionales para representar el estado
# Definimos una función de onda general que depende de las fases
theta_1, theta_2, theta_3 = sp.symbols('theta_1 theta_2 theta_3', real=True)
J = sp.symbols('J', real=True, positive=True)
hbar = sp.symbols('hbar', real=True, positive=True)

# El Hamiltoniano de la red clásica/cuántica simplificado (operador de energía potencial en este espacio)
H_op = J * ((1 - sp.cos(theta_1 - theta_2)) + (1 - sp.cos(theta_2 - theta_3)) + (1 - sp.cos(theta_3 - theta_1)))

# Definimos un operador P(0), por ejemplo, un operador que depende puramente de las coordenadas (fases)
# Si P es un operador de posición/fase (ej. cos(theta_1)), conmutará con el potencial H_op
# ya que ambos son funciones de las coordenadas y los operadores de coordenadas conmutan entre sí: [V(theta), P(theta)] = 0
P_op = sp.cos(theta_1) # Representación de P(0)

# Para cualquier estado base |0> representado por una función de onda psi_0(theta_1, theta_2, theta_3)
# Como H_op y P_op son operadores multiplicativos en la representación de posición/fase:
# H @ P @ |0>  es equivalente a H_op * P_op * psi_0
# P @ H @ |0>  es equivalente P_op * H_op * psi_0

# Evaluamos analíticamente el conmutador [H, P] operando sobre un estado genérico psi_0:
# [H, P] = H_op * P_op - P_op * H_op
conmutador_op = H_op * P_op - P_op * H_op

print("Operador Conmutador [H, P]:")
print(conmutador_op.simplify())

# Verificación en el estado base |0> (donde todos los theta_i = 0)
estado_cero = {theta_1: 0, theta_2: 0, theta_3: 0}
resultado_en_cero = conmutador_op.subs(estado_cero).simplify()

print(f"\nEvaluación del conmutador en el estado de silencio |0>: {resultado_en_cero}")
```

### Generalización para $n$ nodos y análisis de la disparidad

Definiremos una función que toma el número de nodos $n$, las amplitudes originales (asignando valores arbitrarios), calcula las normalizadas, el factor de escala $T$ y compara los productos generales de ambos lados:

* **Lado Izquierdo (Original):** $\prod_{i=1}^{n} X_i$
* **Lado Derecho (Normalizado):** $n \cdot \prod_{i=1}^{n} x_i$

```python
import numpy as np
import sympy as sp

def analizar_disparidad_n_nodos(n, valores_originales=None):
    # Si no se proporcionan valores, se asignan valores de prueba (ej. todos 1 excepto el último que es n-1)
    if valores_originales is None:
        valores_originales = [1.0] * (n - 1) + [float(n - 1)]

    assert len(valores_originales) == n, f"Se requieren exactamente {n} valores."

    # 1. Calcular el total de escala T como la suma de los primeros n-1 elementos
    T = sum(valores_originales[:-1])

    # 2. Calcular variables normalizadas minúsculas
    # Las primeras n-1 son proporciones x_i = X_i / T
    variables_normalizadas = [X / T for X in valores_originales[:-1]]

    # El último elemento normalizado se define análogamente como T menos la suma de las anteriores
    ultimo_normalizado = T - sum(variables_normalizadas)
    variables_normalizadas.append(ultimo_normalizado)

    # 3. Definir la variable de escala C del último elemento (análogo a C = A / a)
    # Usamos el primer elemento como referencia: C = X_1 / x_1
    C = valores_originales[0] / variables_normalizadas[0]

    # Ajustar la lista de variables originales para incluir C en lugar de usar solo valores arbitrarios
    # En el ejemplo original: Lado Izquierdo = A * B * C
    # Generalizado: Producto de todas las variables reemplazando la última por el factor de escala C
    variables_izq = valores_originales[:-1] + [C]

    # Lado izquierdo y derecho genérico
    lado_izq = np.prod(variables_izq)
    # Lado derecho generalizado: n * producto(variables_normalizadas)
    lado_der = n * np.prod(variables_normalizadas)

    diferencia = lado_izq - lado_der

    print(f"=== Análisis para n = {n} nodos ===")
    print(f"Valores originales (X): {valores_originales}")
    print(f"Escala T: {T}")
    print(f"Valores normalizados (x): {variables_normalizadas}")
    print(f"Factor de escala C (X_1 / x_1): {C}")
    print(f"Lado Izquierdo (X_1 * ... * X_(n-1) * C): {lado_izq}")
    print(f"Lado Derecho ({n} * x_1 * ... * x_n): {lado_der}")
    print(f"¿Son iguales?: {np.isclose(lado_izq, lado_der)}")
    print(f"Diferencia absoluta: {abs(diferencia)}\n")

    return lado_izq, lado_der, diferencia

# Probar con n = 3 (debe coincidir con la prueba anterior)
analizar_disparidad_n_nodos(3, [1.0, 1.0, 2.0])

# Probar con n = 4
analizar_disparidad_n_nodos(4, [1.0, 1.0, 1.0, 3.0])

# Probar con n = 5
analizar_disparidad_n_nodos(5, [1.0, 1.0, 1.0, 1.0, 4.0])
```

### Análisis Simbólico: Relación entre la Disparidad $D$ y el Ángulo/Proporción del Primer Nodo

Definiremos las variables de forma completamente algebraica usando `sympy` para obtener la ecuación exacta de $D$ en función de la proporción del primer nodo $a$ (que actúa como el ángulo normalizado de crecimiento del nodo 1) y el factor de escala $T$.

```python
import sympy as sp

# Definir variables simbólicas
# T: Escala total (A + B)
# a: proporción del primer nodo (representa su contribución angular / fase relativa)
T = sp.symbols('T', real=True, positive=True)
a = sp.symbols('a', real=True, positive=True)

# Bajo la parametrización de 3 nodos:
# A = a * T
# b = B / T. Como a + b = 1 (para n=3), entonces b = 1 - a
# B = b * T = (1 - a) * T
# C = A / a = T
# c = T - (a + b) = T - 1

A_sym = a * T
b_sym = 1 - a
B_sym = b_sym * T
C_sym = T
c_sym = T - 1

# Lado Izquierdo (Original)
lado_izq_sym = A_sym * B_sym * C_sym

# Lado Derecho (Normalizado)
lado_der_sym = 2 * a * b_sym * c_sym

# Diferencia o Disparidad D
D_sym = lado_izq_sym - lado_der_sym

print("Expresión simbólica de la Diferencia D en función de T y el ángulo/proporción 'a':")
sp.pprint(sp.factor(D_sym))

print("\nPara nuestro caso de estudio (T = 2):")
D_caso_T2 = D_sym.subs(T, 2).simplify()
sp.pprint(D_caso_T2)

# Resolver para qué valor de 'a' (ángulo) la disparidad es máxima
derivada_D = sp.diff(D_caso_T2, a)
a_max = sp.solve(derivada_D, a)
print(f"\nEl valor del 'ángulo/proporción' del primer nodo (a) que maximiza la disparidad es: {a_max}")
```

```python
# Definir el factor de escala homogéneo lambda
lambda_val = 2

# Escalar las amplitudes originales
A_p = lambda_val * A
B_p = lambda_val * B

# Calcular el nuevo total de escala T_p
T_p = A_p + B_p

# Recalcular las variables normalizadas (las proporciones deben mantenerse invariantes)
a_p = A_p / T_p
b_p = B_p / T_p

# Calcular los nuevos factores de escala correspondientes
C_p = A_p / a_p
c_p = T_p - (a_p + b_p)

# Evaluar los nuevos lados de la relación de disparidad
lado_izq_p = A_p * B_p * C_p
lado_der_p = 2 * a_p * b_p * c_p
diferencia_p = lado_izq_p - lado_der_p

print("=== ESCALAMIENTO DEL SISTEMA POR LAMBDA = 2 ===")
print(f"Nuevos valores originales: A' = {A_p}, B' = {B_p}")
print(f"Nueva escala total T': {T_p}")
print(f"Nuevas proporciones (deben ser idénticas): a' = {a_p}, b' = {b_p}")
print(f"Nuevos factores de escala individuales: C' = {C_p}, c' = {c_p}")
print(f"Nuevo Lado Izquierdo (Original escalado): {lado_izq_p}")
print(f"Nuevo Lado Derecho (Normalizado escalado): {lado_der_p}")
print(f"Nueva Disparidad D': {diferencia_p}")
print(f"Relación de amplificación de disparidad (D' / D): {diferencia_p / diferencia}")
```

```python
import sympy as sp
import numpy as np

# 1. Definir los parámetros de la red de 19 nodos
N = 19
J = sp.symbols('J', real=True, positive=True)

# Crear fases simbólicas para cada uno de los 19 nodos
thetas = sp.symbols(f'theta_1:{N+1}', real=True)

# 2. Definir la topología de la red hexagonal centrada (Matriz de adyacencia de 19x19)
# Capa 0 (Central): Nodo 1 (índice 0)
# Capa 1 (Anillo interior): Nodos 2 a 7
# Capa 2 (Anillo exterior): Nodos 8 a 19

adj_matrix = np.zeros((N, N))

# Conexiones del centro (nodo 1) a la primera capa (nodos 2 a 7)
for i in range(1, 7):
    adj_matrix[0, i] = 1
    adj_matrix[i, 0] = 1

# Conexiones de anillo en la primera capa
for i in range(1, 7):
    siguiente = 1 + (i % 6)
    adj_matrix[i, siguiente] = 1
    adj_matrix[siguiente, i] = 1

# Conexiones de la primera capa (2-7) a la segunda capa (8-19)
# En un hexágono centrado, cada uno de los 6 nodos de la primera capa se conecta a nodos específicos de la capa exterior
for i in range(6):
    nodo_c1 = 1 + i
    # Esquinas exteriores
    nodo_c2_esq = 7 + 2 * i
    # Lados exteriores
    nodo_c2_lado1 = 7 + (2 * i + 1) % 12
    nodo_c2_lado2 = 7 + (2 * i - 1) % 12

    adj_matrix[nodo_c1, nodo_c2_esq] = 1
    adj_matrix[nodo_c2_esq, nodo_c1] = 1

    adj_matrix[nodo_c1, nodo_c2_lado1] = 1
    adj_matrix[nodo_c2_lado1, nodo_c1] = 1

# Conexiones de anillo en la segunda capa (anillo de 12 nodos)
for i in range(12):
    nodo_actual = 7 + i
    siguiente = 7 + (i + 1) % 12
    adj_matrix[nodo_actual, siguiente] = 1
    adj_matrix[siguiente, nodo_actual] = 1

# 3. Construir el Hamiltoniano de la red hexagonal para 19 nodos
H_hex = 0
for i in range(N):
    for j in range(i + 1, N):
        if adj_matrix[i, j] == 1:
            H_hex += (1 - sp.cos(thetas[i] - thetas[j]))
H_hex = J * H_hex

print("=== CONFIGURACIÓN DE RED HEXAGONAL DE 19 NODOS ===")
print(f"Número total de enlaces activos (coordinación): {int(np.sum(adj_matrix) / 2)}")
print(f"Estado de Silencio (todas las fases theta = 0):")
silencio_19 = {theta: 0 for theta in thetas}
print(f"H_hex(theta=0) = {H_hex.subs(silencio_19)}")

# 4. Calcular disparidad generalizada para n = 19 con valores base unitarios
print("\n=== EVALUACIÓN DE DISPARIDAD ASINTÓTICA PARA N = 19 ===")
izq_19, der_19, dif_19 = analizar_disparidad_n_nodos(19)
```

```python
import numpy as np

# 1. Configurar la semilla para reproducibilidad
np.random.seed(42)

# 2. Definir fases aleatorias en el rango [-pi, pi] para los 57 nodos
fases_aleatorias_val = np.random.uniform(-np.pi, np.pi, N_57)

# Mapear cada variable simbólica de fase a su valor aleatorio correspondiente
estado_aleatorio = {thetas_57[i]: fases_aleatorias_val[i] for i in range(N_57)}

# 3. Construir la matriz de adyacencia de la red hexagonal extendida para calcular H
# Para N=57, podemos modelar las conexiones de forma sistemática o aproximar una estructura de red hexagonal regular.
# Generemos una matriz de adyacencia aleatoria simétrica pero dispersa de 57x57 para representar la topología de la red física
# con un grado de coordinación promedio de ~4 por nodo (similar a redes de grafeno/hexagonales con bordes).

adj_57 = np.zeros((N_57, N_57))
# Conectamos secuencialmente en anillo para asegurar conectividad básica
for i in range(N_57):
    siguiente = (i + 1) % N_57
    adj_57[i, siguiente] = 1
    adj_57[siguiente, i] = 1

# Agregar conexiones hexagonales concéntricas adicionales (puentes de capas interconectadas)
for i in range(N_57 // 3):
    # Conectar nodos de capas internas y externas de manera regular
    idx_interno = i
    idx_externo = (i + 19) % N_57
    adj_57[idx_interno, idx_externo] = 1
    adj_57[idx_externo, idx_interno] = 1

# 4. Construir y evaluar el Hamiltoniano H_57 de manera simbólica y luego numérica
H_simb_57 = 0
for i in range(N_57):
    for j in range(i + 1, N_57):
        if adj_57[i, j] == 1:
            H_simb_57 += (1 - sp.cos(thetas_57[i] - thetas_57[j]))

H_evaluado_57 = H_simb_57.subs(estado_aleatorio).evalf()

print("=== VERIFICACIÓN DEL HAMILTONIANO CON FASES ALEATORIAS (N = 57) ===")
print(f"Número total de enlaces activos definidos en la topología: {int(np.sum(adj_57) / 2)}")
print(f"Primeras 5 fases aleatorias (en radianes): {[round(float(f), 4) for f in fases_aleatorias_val[:5]]}")
print(f"Valor de expectación numérico del Hamiltoniano (H_57) = {H_evaluado_57} * J_57")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Parámetros de la simulación
num_simulaciones = 10000
energias_calculadas = []

# Obtener los índices de los enlaces activos a partir de la matriz de adyacencia adj_57
enlaces = np.argwhere(np.triu(adj_57, k=1) == 1)

# 2. Simulación de Monte Carlo
for _ in range(num_simulaciones):
    # Generar fases aleatorias en el rango [-pi, pi] para los 57 nodos
    fases = np.random.uniform(-np.pi, np.pi, N_57)

    # Calcular numéricamente el Hamiltoniano usando las fases
    # H = sum_{i < j} (1 - cos(theta_i - theta_j)) para cada enlace activo
    diff_fases = fases[enlaces[:, 0]] - fases[enlaces[:, 1]]
    H_val = np.sum(1 - np.cos(diff_fases))
    energias_calculadas.append(H_val)

# Convertir a arreglo de numpy para estadísticas
energias_calculadas = np.array(energias_calculadas)

# 3. Graficar la distribución
plt.figure(figsize=(10, 6))
sns.histplot(energias_calculadas, kde=True, color='teal', edgecolor='black', bins=50, stat='density')

# Líneas de referencia
plt.axvline(np.mean(energias_calculadas), color='red', linestyle='--', linewidth=2,
            label=f'Energía Promedio: {np.mean(energias_calculadas):.2f} * J_57')
plt.axvline(0, color='darkgreen', linestyle='-', linewidth=2,
            label='Estado de Silencio (Mínima Energía = 0)')
plt.axvline(len(enlaces), color='purple', linestyle=':', linewidth=2,
            label=f'Energía Máxima Teórica (Enlaces = {len(enlaces)})')

plt.title('Distribución de Energías del Hamiltoniano ($H_{57}$) para 10k Configuraciones Aleatorias', fontsize=13)
plt.xlabel('Energía del Hamiltoniano ($H_{57} / J_{57}$)', fontsize=11)
plt.ylabel('Densidad de Probabilidad', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(fontsize=10)
plt.xlim(0, len(enlaces) + 5)
plt.show()

# Mostrar estadísticas descriptivas
print("=== ESTADÍSTICAS DE LA DISTRIBUCIÓN DE ENERGÍA ===")
print(f"Energía mínima observada: {np.min(energias_calculadas):.4f} * J_57")
print(f"Energía promedio: {np.mean(energias_calculadas):.4f} * J_57")
print(f"Energía máxima observada: {np.max(energias_calculadas):.4f} * J_57")
print(f"Desviación estándar: {np.std(energias_calculadas):.4f} * J_57")
```
